# NumPy: modelo armónico y razonamiento sobre arreglos

Objetivos: construir arreglos, inspeccionar forma y tipo, aplicar operaciones vectorizadas y crear una figura con la interfaz orientada a objetos de Matplotlib.

```python
import matplotlib.pyplot as plt
import numpy as np
```

Antes de ejecutar la siguiente celda, prediga la forma, el número de elementos y los valores inicial y final de `tiempo_s`.

```python
tiempo_s = np.linspace(0.0, 4.0, 401)
amplitud_m = 0.08
frecuencia_hz = 1.0
omega_rad_s = 2 * np.pi * frecuencia_hz
posicion_m = amplitud_m * np.cos(omega_rad_s * tiempo_s)

print(tiempo_s.shape, tiempo_s.dtype)
print(posicion_m.min(), posicion_m.max())
```

La expresión matemática actúa elemento a elemento. No hay un ciclo explícito en el código de usuario, aunque NumPy sí debe recorrer los datos internamente.

```python
mascara = posicion_m > 0
print(f"Fracción de muestras positivas: {mascara.mean():.3f}")
print("Primeros tiempos con x > 0:", tiempo_s[mascara][:5])
```

```python
fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(tiempo_s, posicion_m, label="x(t)")
ax.set(xlabel="Tiempo (s)", ylabel="Posición (m)", title="Oscilador armónico")
ax.axhline(0.0, color="0.5", linewidth=0.8)
ax.legend()
fig.tight_layout()
```
