# Auditoría antes de limpiar
No elimine filas hasta formular y justificar una regla.

```python
import pandas as pd
tabla = pd.DataFrame({'tiempo_s': [0, 1, 1, 3], 'valor': [2.1, None, None, 2.8], 'sensor': ['A', 'A', 'A', 'B']})
tabla
```

```python
auditoria = {
    'filas': len(tabla),
    'faltantes': tabla.isna().sum().to_dict(),
    'duplicadas': int(tabla.duplicated().sum()),
}
auditoria
```
