```python
import matplotlib.pyplot as plt
import numpy as np
x = np.linspace(0, 2*np.pi, 100)
y = np.sin(x)
```

```python
fig, ax = plt.subplots(layout='constrained')
ax.plot(x, y, label='modelo')
ax.set(xlabel='Fase [rad]', ylabel='Amplitud [u.a.]')
ax.legend();
```
